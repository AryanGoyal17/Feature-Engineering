```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df_date = pd.read_csv("../data/orders.csv")
df_time = pd.read_csv("../data/messages.csv")
```

```python
# By default -- date and time features are treated as str/object dtype
```

```python
## Converting dtype from str to datetime of orders data ---

df_date['date'] = pd.to_datetime(df_date['date'])
```

```python
df_date.info()
```

**Extract year**

```python
df_date['year'] = df_date['date'].dt.year
```

```python
df_date.head()
```

**Extract Month Number and Month Name**

```python
df_date['month_number'] = df_date['date'].dt.month
```

```python
df_date.head()
```

```python
df_date['month_name'] = df_date['date'].dt.month_name()
```

```python
df_date
```

**Extract Day no (of month), Day no(of week), Day name**

```python
df_date['day_number'] = df_date['date'].dt.day
```

```python
df_date.head()
```

```python
df_date['day_number_of_week'] = df_date['date'].dt.dayofweek
```

```python
df_date.head()
```

```python
df_date['day_name'] = df_date['date'].dt.day_name()
```

```python
df_date.head()
```

**Is it a weekend or not?**

```python
df_date['is_weekend'] = np.where(df_date['day_name'].isin(['Saturday', 'Sunday']), 1, 0)
```

```python
df_date.head()
```

**Which week of the year?**

```python
df_date['week_no'] = df_date['date'].dt.isocalendar().week
```

```python
df_date.head()
```

**Extract quarter**

```python
df_date['quarter_no'] = df_date['date'].dt.quarter
```

```python
df_date.head()
```

**Extract semester**

```python
df_date['semester'] = np.where(df_date['quarter_no'].isin([1,2]), 2, 1)
```

```python
df_date.head()
```

**Extract time elapsed between dates**

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - df_date['date']
```

```python
# No of days elapsed --

(today - df_date['date']).dt.days
```

```python
# No of months passed --

np.round((today - df_date['date']) / np.timedelta64(30, 'D'), 0)
```

```python
# Converting the dtype from string to datetime

df_time['date'] = pd.to_datetime(df_time['date'])
```

**Extracting hours, minutes and seconds**

```python
df_time['hour'] = df_time['date'].dt.hour
df_time['minutes'] = df_time['date'].dt.minute
df_time['seconds'] = df_time['date'].dt.second
```

```python
df_time.head()
```

**Extracting only the time portion**

```python
df_time['time'] = df_time['date'].dt.time
```

```python
df_time.head()
```

**Extracting time elapsed**

```python
today - df_time['date']
```

```python
# seconds

np.round((today - df_time['date']) / np.timedelta64(1, 's'), 0)
```

```python
# minutes

np.round((today - df_time['date']) / np.timedelta64(1, 'm'), 0)
```

```python
# hours

np.round((today - df_time['date']) / np.timedelta64(1, 'h'), 0)
```
