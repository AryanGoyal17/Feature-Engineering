```python
upper_limit = df['Height'].quantile(0.99)
lower_limit = df['Height'].quantile(0.01)
```

```python
upper_limit, lower_limit
```

# Trimming

```python
df_trimming = df[(df['Height'] > lower_limit) & (df['Height'] < upper_limit)]
df_trimming
```

```python
sns.histplot(df_trimming['Height'], kde=True)
```

```python
sns.boxplot(df_trimming['Height'])
```

```python
df_trimming.describe()
```

# Capping (Also called Winsorization)

```python
df['Height'] = np.where(
    df['Height'] > upper_limit,
    upper_limit,
    np.where(
        df['Height'] < lower_limit,
        lower_limit,
        df['Height']
    )
)
```

```python
df.describe()
```

```python
sns.histplot(df['Height'], kde=True)
```

```python
sns.boxplot(df['Height'])
```
